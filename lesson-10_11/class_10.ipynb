```python
URL = "https://www.basketball-reference.com/leagues/NBA_2022_games.html"
FILENAME = "basket.csv"

def load_dataframe(url, filename):
    if os.path.exists(filename):
        print("load from file", filename)
        df = pd.read_csv(filename)
    else:
        print("load from url")
        data = pd.read_html(url)
        df = data[0]
        df.to_csv(filename)

    return df


df = load_dataframe(URL, FILENAME)
```
