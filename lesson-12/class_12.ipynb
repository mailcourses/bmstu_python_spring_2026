```python
# O(n * m) - time
# O(1) - memory

def inteception_naive(nums1: list[int], nums2: list[int]) -> list[int]:
    res = []
    for n in nums1:
        if n in nums2:
            res.append(n)
    return res


assert inteception_naive([1, 3, 5, 7], [2, 3, 6, 7, 22]) == [3, 7]
assert inteception_naive([1, 3, 5, 7], [2, 4, 6, 22]) == []
```

```python
# O(n + m + min(n, m) + min(n, m)) ~ O(n + m) - time
# O(n + m) - memory

def inteception_set(nums1: list[int], nums2: list[int]) -> list[int]:
    return list(set(nums1) & set(nums2))


assert inteception_set([1, 3, 5, 7], [2, 3, 6, 7, 22]) == [3, 7]
assert inteception_set([1, 3, 5, 7], [2, 4, 6, 22]) == []
```

```python
# O(n * log m) - time
# O(1) - memory

def inteception_bisect(nums1: list[int], nums2: list[int]) -> list[int]:
    res = []
    for n in nums1:
        pos = bisect.bisect_left(nums2, n)
        if pos != len(nums2) and nums2[pos] == n:
            res.append(n)
    return res


assert inteception_bisect([1, 3, 5, 7], [2, 3, 6, 7, 22]) == [3, 7]
assert inteception_bisect([1, 3, 5, 7], [2, 4, 6, 22]) == []
```

```python
assert bisect.bisect_left([1, 2, 2, 3, 4], 0) == 0
assert bisect.bisect_left([1, 2, 2, 3, 4], 1) == 0
assert bisect.bisect_left([1, 2, 2, 3, 4], 2) == 1
assert bisect.bisect_left([1, 2, 2, 3, 4], 5) == 5
```

```python
# O(n + m) - time
# O(1) - memory

def inteception_iter(nums1: list[int], nums2: list[int]) -> list[int]:
    i1 = 0
    i2 = 0
    res = []
    
    while i1 < len(nums1) and i2 < len(nums2):
        if ( (el1 := nums1[i1]) == (el2 := nums2[i2]) ):
            res.append(el1)
            i1 += 1
            i2 += 1
        elif el1 < el2:
            i1 += 1
        else:
            i2 += 1

    return res


assert inteception_iter([1, 3, 5, 7], [2, 3, 6, 7, 22]) == [3, 7]
assert inteception_iter([1, 3, 5, 7], [2, 4, 6, 22]) == []
```

```python
# O(n) - time
# O(m) - memory
# unstable


def counter_sort(nums: list[int]) -> list[int]:
    m = max(nums)  # O(1)
    counter = [0] * (m + 1)

    for n in nums:
        counter[n] += 1

    print(f"{counter=}")

    res = []
    for n, cnt in enumerate(counter):
        for _ in range(cnt):
            res.append(n)

    return res


assert counter_sort(lst) == sorted(lst)
assert counter_sort([4, 3, 5, 1, 1, 1]) == sorted([4, 3, 5, 1, 1, 1])
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"P[{self.name}, {self.age}]"

    def __repr__(self):
        return f"P[{self.name}, {self.age}]"


persons = [
    Person("bob", 10),
    Person("steve", 20),
    Person("tom", 10),
    Person("peter", 10),
    Person("alex", 20),
]
```

```python
# O(n + m) - time
# O(m) - memory
# stable


def counter_sort_stable(nums: list[int]) -> list[int]:
    m = max(nums)  # O(1)
    counter = [0] * (m + 1)

    for n in nums:
        counter[n] += 1

    print(f"init {counter=}")

    for i in range(1, len(counter)):
        counter[i] += counter[i - 1]

    print(f"part {counter=}")

    res = [0] * len(nums)

    for i in range(len(nums) - 1, -1, -1):
        n = nums[i]
        pos = counter[n] - 1
        res[pos] = n
        counter[n] -= 1

    return res


assert counter_sort_stable(lst) == sorted(lst)
assert counter_sort_stable([4, 3, 5, 1, 1, 1]) == sorted([4, 3, 5, 1, 1, 1])
```
