```python
class DataProcess(threading.Thread):
    def __init__(self, n):
        super().__init__()
        self.n = n
        self.result = 0

    def run(self):
        print(f"DataProcess.run {self.name=}, {self.n=}, {self.ident=}")

        for i in range(self.n):
            print(f"iter {i=} {self.n=}")
            self.result += i
            time.sleep(1)

        print(f"DataProcess.run finished {self.name=}, {self.n=}, {self.result=}")
```

```python
%%time

N = 50
N_THREADS = 10
N_JOBS = N // N_THREADS

URL = "https://wiki.python.org/moin/BeginnersGuide"
URLS = [URL] * N


def fetch_url(urls):
    t1 = time.time()
    print(f"{threading.current_thread().name=} start", len(urls))

    res = []
    for i, url in enumerate(urls):
        # data = urlopen(url)
        # res.append(data.code)
        # print(f"{threading.current_thread().name=} open {i=}")
        time.sleep(random.random())

    print(f"{threading.current_thread().name=} finish, {res=}", time.time() - t1)
    return res


threads = [
    threading.Thread(
        target=fetch_url,
        name=f"fetch_{i}",
        args=(URLS[i * N_JOBS : (i + 1) * N_JOBS],),
    )
    for i in range(N_THREADS)
]
for th in threads:
    th.start()

for th in threads:
    th.join()
```

```python
%%time

N = 50
N_THREADS = 10
N_JOBS = N // N_THREADS

URL = "https://wiki.python.org/moin/BeginnersGuide"
URLS = [URL] * N


def fetch_urls(que):
    t1 = time.time()
    print(f"{threading.current_thread().name=} start")
    cnt = 0

    while True:
        url = que.get()
        if url is None:
            que.put(url)
            break

        # data = urlopen(url)
        # res.append(data.code)
        # print(f"{threading.current_thread().name=} open {i=}")
        time.sleep(random.random())
        cnt += 1

    print(f"{threading.current_thread().name=} finish {cnt=}", time.time() - t1)


def run():
    que = queue.Queue()
    for url in URLS:
        que.put(url)

    que.put(None)

    threads = [
        threading.Thread(
            target=fetch_urls,
            name=f"fetch_{i}",
            args=(que,),
        )
        for i in range(N_THREADS)
    ]
    for th in threads:
        th.start()
    
    for th in threads:
        th.join()


run()
```

```python
import multiprocessing as mp
import socket
```

```python
mp_ctx = mp.get_context("fork")
```

```python
def server():
    print("start server")
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
        sock.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)

        sock.bind(("localhost", 20_000))
        sock.listen(5)

        i = 5
        while i > 0:
            i -= 1
            
            client_sock, address = sock.accept()
            print("server accepted client", address)

            data = client_sock.recv(1048)
            data = data.decode().upper()
            client_sock.sendall(data.encode())
            print("server finished to client", address)

    print("server finished")


def client(data):
    print("start client")
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
        sock.connect(("localhost", 20_000))

        sock.sendall(data.encode())

        data = sock.recv(1048).decode()
        print("client recv from server", data)

    print("client finished")


proc_server = mp_ctx.Process(target=server)
proc_client = mp_ctx.Process(target=client, args=("кружка",))

proc_server.start()
time.sleep(1)
proc_client.start()

# proc_server.join()
proc_client.join()

print("total finish")
```
