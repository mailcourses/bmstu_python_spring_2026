```python
class Person:
    def __init__(self, name):
        print(f"Person {name=}", self)
        self.name = name


class Student(Person):
    def __init__(self, school, *args, **kwargs):
        print(f"Student {school=}", self)
        super().__init__(*args, **kwargs)
        self.school = school


class Teacher(Person):
    def __init__(self, course, *args, **kwargs):
        print(f"Teacher {course=}", self)
        super().__init__(*args, **kwargs)
        self.course = course


class PostGraduate(Student, Teacher):
    def __init__(self, degree, *args, **kwargs):
        print(f"PostGraduate {degree=}", self)
        super().__init__(*args, **kwargs)
        self.degree = degree

    def eat(self, food):
        print(f"{food} is eaten")
```

```python
import cProfile
import pstats
import io
```

```python
prof = cProfile.Profile()
prof.enable()

steve = PostGraduate("phd", "mstu", "python", "steve")
steve.eat("apple")

prof.disable()
```

```python
s = io.StringIO()
stats = pstats.Stats(prof, stream=s).sort_stats(pstats.SortKey.CUMULATIVE)
stats.print_stats()

print(s.getvalue())
```

```python
def create():
    steve = PostGraduate("phd", "mstu", "python", "steve")
    steve.eat("apple")


prof = cProfile.Profile()
prof.runcall(create)

print(prof)


s = io.StringIO()
stats = pstats.Stats(prof, stream=s).sort_stats(pstats.SortKey.CUMULATIVE)
stats.print_stats()

print(s.getvalue())
```

```python
class CustomFilter(logging.Filter):
    def filter(self, rec):
        print(rec.args)
        return rec.args[0] != "steve"


def get_logger(name):
    logger = logging.getLogger(name)
    logger.setLevel(logging.INFO)

    while logger.handlers:
        logger.removeHandler(logger.handlers[0])

    handler = logging.StreamHandler()
    handler.setLevel(logging.INFO)
    handler.setFormatter(logging.Formatter(f"EDX-2 %(asctime)s\t%(levelname)s\t%(name)s\t%(message)s"))

    handler.addFilter(CustomFilter())

    logger.addHandler(handler)
    logger.propagate = False

    return logger
```
