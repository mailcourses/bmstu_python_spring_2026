```python
def gen_fib(n):
    print("start fib")
    a, b = 0, 1
    while n > 0:
        print("before yield", b)
        yield b
        print("after yield", b)
        a, b = b, a + b
        n -= 1
    return "all fibs are done"
```

```python
for x in gen_fib(5):
    print(x)
```

```python
def counter_gen(start, stop, step=1):
    print(f"launch counter_gen({start=}, {stop=}, {step=})")
    while start < stop:
        yield start
        start += step
```

```python
for i in counter_gen(10, 15):
    print(f"{i=}")
```

```python
for i in counter_gen(10, 15, 3):
    print(f"{i=}")
```

```python
def counter_gen(start, stop, step=1):
    print(f"launch counter_gen({start=}, {stop=}, {step=})")
    while start < stop:
        step_candidad = yield start
        print(f"{step_candidad=}")
        if step_candidad is not None:
            step = step_candidad

        start += step
```

```python
counter = counter_gen(10, 20, 2)
```

# Пример наследования

```python
class Person:
    def __init__(self, name):
        print(f"Person {name=}", self)
        self.name = name


class Student(Person):
    def __init__(self, school, *args, **kwargs):
        print(f"Student {school=}", self)
        super().__init__(*args, **kwargs)
        self.school = school


class Teacher(Person):
    def __init__(self, course, *args, **kwargs):
        print(f"Teacher {course=}", self)
        super().__init__(*args, **kwargs)
        self.course = course


class PostGraduate(Student, Teacher):
    def __init__(self, degree, *args, **kwargs):
        print(f"PostGraduate {degree=}", self)
        super().__init__(*args, **kwargs)
        self.degree = degree
```

```python
steve = PostGraduate("phd", "mstu", "python", "steve")
```
